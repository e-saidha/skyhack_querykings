```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

flight = pd.read_csv("/content/Flight Level Data.csv")
pnr = pd.read_csv("/content/PNR Remark Level Data.csv")
pnr_rem = pd.read_csv("/content/PNR Remark Level Data.csv")
bag = pd.read_csv("/content/Bag+Level+Data.csv")
airports = pd.read_csv("/content/Airports Data.csv")

datasets = {
    "Flight Level": flight,
    "PNR Flight": pnr,
    "PNR Remarks": pnr_rem,
    "Bag Level": bag,
    "Airports": airports
}

all_columns = set()
for df in datasets.values():
    all_columns.update(df.columns)
all_columns = sorted(list(all_columns))

missing_matrix = pd.DataFrame(0,
                              index=list(datasets.keys()),
                              columns=all_columns)

for dataset_name, df in datasets.items():
    for col in df.columns:
        missing_count = df[col].isnull().sum()
        missing_matrix.loc[dataset_name, col] = missing_count

plt.figure(figsize=(20, 8))
sns.heatmap(missing_matrix,
            annot=True,
            fmt='g',
            cmap='Reds',
            cbar_kws={'label': 'Missing Values'},
            linewidths=0.5,
            linecolor='lightgray')

plt.title('Data Quality — Missing Values Across Datasets',
          fontsize=16,
          fontweight='bold',
          pad=20)
plt.xlabel('Columns', fontsize=12, fontweight='bold')
plt.ylabel('Datasets', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

print("MISSING VALUES SUMMARY")
for name, df in datasets.items():
    total_missing = df.isnull().sum().sum()
    total_cells = df.shape[0] * df.shape[1]
    missing_pct = (total_missing / total_cells) * 100
    print(f"\n{name}:")
    print(f"  Total missing values: {total_missing}")
    print(f"  Missing percentage: {missing_pct:.2f}%")
    print(f"  Columns with missing values: {(df.isnull().sum() > 0).sum()}")

print("DUPLICATE ROWS SUMMARY")
for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f"{name}: {duplicates} duplicate rows")
```
